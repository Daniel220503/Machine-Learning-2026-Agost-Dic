####*Simple linear regression measures the relationship between an independent variable on a dependent variable and measures the correlation between them.*


### Mathematical Model and Implementation

# Simple Linear Regression in Python

In statistics, simple linear regression is a linear regression model with a single explanatory variable. In simple linear regression, we predict scores on one variable based on results on another. The criteria variable Y is the variable we are predicting. Predictor variable X is the variable using which we are making our predictions.

---

## Mathematical Model and Implementation

### Equation : y = mx + c
This is the simple linear regression equation where c is the constant and m is the slope and describes the relationship between x (independent variable) and y (dependent variable)


$Y_i = \beta_0 + \beta_1 X_i$

####  Prediction Equation
$$y_{\text{pred}} = w \cdot x + b$$

* **$x$**: Independent Variable
* **$y_{\text{pred}}$**: Predicted Value (`y_pred`)
* **$w$**: Slope (`w = 0.8`)
* **$b$**: y-Intercept (`b = 1.0`)

#### Expression for the Predicted Value
$$y_{\text{pred}}^{(i)} = w \cdot x_i + b$$

#### 2. Loss Function (MSE)
$$\text{MSE} = \frac{1}{n} \sum_{i=1}^{n} (y_{\text{pred}}^{(i)} - y_{\text{true}}^{(i)})^2$$

* **$y_{\text{true}}^{(i)}$**: Actual values (`y_true`)
* **$y_{\text{pred}}^{(i)}$**: Predicted values (`y_pred`)
* **$n$**: Total number of samples (`len(x)`)

#Python Implementation of Simple Linear Regression

####Import the library of python as np

```python
import numpy as np
```

####We define the prediction function, this is where we going to introduce our independent variable values to make the predictions.

```python
def predict(x, w, b):
    return w * x + b
```

####Definition of the loss fuction, this perfomrs the calcule to define the error of the prediction

```python
def mse_loss(y_pred, y_true):
    return np.mean((y_pred - y_true) ** 2)
    #We call the method mean of the library numpy to calculate the loss value for us
```

####Example data

```python
x = np.array([1, 2, 3, 4, 5])  #Independent variable as a unidimensional numpy array
y_true = np.array([2, 4, 5, 4, 5])  #Dependent variable as a unidimensional numpy array
```

####Model parameters (slope and intercept)

```python
w = 0.8
b = 1.0
```

####We call the prediction function and set the result value in a variable

```python
y_pred = predict(x, w, b)
```

####We do the same for the loss function

```python
loss = mse_loss(y_pred, y_true)
```

####Now we just print the results

```python
print(f"Predicted values: {y_pred}")
print(f"MSE Loss: {loss:.3f}")

# Predicted values: [1.8 2.6 3.4 4.2 5.]
# MSE Loss: 0.920
```
